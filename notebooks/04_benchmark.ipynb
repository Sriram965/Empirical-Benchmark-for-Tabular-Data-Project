```python
import sys
sys.path.append('..')
import numpy as np
import optuna
optuna.logging.set_verbosity(optuna.logging.WARNING)

from src.tree_models import get_random_forest, get_hist_gradient_boosting, get_xgboost
from src.dl_models import get_mlp, get_resnet, get_ft_transformer

input_dim, output_dim = 10, 2
X = np.random.randn(200, input_dim).astype(np.float32)
y = np.random.randint(0, 2, 200).astype(np.int64)

study = optuna.create_study(direction='maximize')

def test_all_six(trial):
    models = {
        'RandomForest':         get_random_forest(trial, 'classification'),
        'HistGradientBoosting': get_hist_gradient_boosting(trial, 'classification'),
        'XGBoost':              get_xgboost(trial, 'classification'),
        'MLP':                  get_mlp(trial, 'classification', input_dim, output_dim),
        'ResNet':               get_resnet(trial, 'classification', input_dim, output_dim),
        'FTTransformer':        get_ft_transformer(trial, 'classification', input_dim, output_dim),
    }
    for name, model in models.items():
        model.fit(X, y)
    return 1.0

study.optimize(test_all_six, n_trials=1)
print("\nAll 6 models working correctly.")
```
